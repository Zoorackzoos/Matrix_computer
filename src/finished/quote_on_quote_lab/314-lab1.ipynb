```python
from IPython.display import display, Math
import numpy as np  ### a python package for linear algebra
import imageio      ### for image file processing
import matplotlib.pyplot as plt

### we give a way to render matrices beautifully:
def show_matrix(A):
    rows = [" & ".join(map(str,row)) for row in A]
    latex = r"\begin{bmatrix}" + r"\\ ".join(rows) + r"\end{bmatrix}"
    display(Math(latex))

### numpy doesn't have a built-in rref function, so we implement it ourselves:
def rref(A, tol=1e-12):
    A = A.astype(float).copy()
    rows, cols = A.shape
    r = 0  # pivot row

    for c in range(cols):
        if r >= rows:
            break

        # Choose pivot with largest absolute value in column
        pivot = np.argmax(np.abs(A[r:, c])) + r

        if abs(A[pivot, c]) < tol:
            continue  # column is effectively zero

        # Swap rows
        A[[r, pivot]] = A[[pivot, r]]

        # Normalize pivot row
        A[r] = A[r] / A[r, c]

        # Eliminate other rows
        for i in range(rows):
            if i != r:
                A[i] = A[i] - A[i, c] * A[r]

        r += 1

    # Clean up tiny floating point artifacts
    A[np.abs(A) < tol] = 0

    return A
```

**Why doesn’t NumPy include a function for RREF?**

So far in this course, **reduced row-echelon form (RREF)** has been our main computational tool. It is natural to wonder why the Python library **NumPy** does not include a built-in `rref` function.

Row reduction is very useful for learning concepts such as rank, linear independence, and solving systems of equations. However, when computers solve large numerical problems, they usually avoid Gauss–Jordan elimination. 
Instead they use other matrix factorizations that are faster and more numerically stable when working with floating-point arithmetic.

For example, numerical libraries often rely on methods like:

* **LU factorization** for solving systems of equations,
* **QR factorization** for least-squares problems (linear regression), and
* **Singular Value Decomposition (SVD)** for tasks like data analysis and image compression.

These methods are better suited to large matrices and the rounding errors that occur with floating-point numbers. 
Because of this, NumPy focuses on providing tools for these widely used numerical algorithms rather than RREF. In this lab we will implement our own `rref` function.
